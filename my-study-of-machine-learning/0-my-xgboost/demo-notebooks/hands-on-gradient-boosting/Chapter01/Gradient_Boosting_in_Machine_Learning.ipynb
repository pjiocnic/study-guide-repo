## 📊 What Values Does `df.describe()` Return?

The `describe()` method in pandas provides summary statistics of the DataFrame’s numeric columns by default.

### Summary of Returned Values:

- **Count**  
  The number of **non-missing (non-NaN)** values in the column.

- **Mean**  
  The **average** value of the column.  
  Calculated as:
  $$
  \text{Mean} = \frac{\sum x_i}{n}
  $$
- **Standard Deviation**  
  A measure of how **spread out** the values are around the mean.  
  Calculated using **sample standard deviation**:  
$$
  \text{Std} = \sqrt{\frac{\sum (x_i - \bar{x})^2}{n - 1}}
$$

- **Minimum**  
  The **smallest** value in the column.

- **Maximum**  
  The **largest** value in the column.

- **Percentiles**  
  Values that divide the **ordered data** into 100 equal parts.  
  By default, `describe()` returns:
  - **25th percentile (Q1)** – lower quartile
  - **50th percentile (Q2)** – median
  - **75th percentile (Q3)** – upper quartile

**Drop Leakage variables**

### ❓ Why Drop `casual` and `registered` Columns When Predicting `cnt`?

```markdown
According to the dataset, the target variable `cnt` (total bike rentals) is the **sum** of:

- `casual` → number of casual (unregistered) users
- `registered` → number of registered users

That is:
```
cnt = casual + registered
```

Including both `casual` and `registered` as input features would result in a model that **can trivially predict `cnt` with 100% accuracy**, because the answer is literally encoded in the inputs.

### 🧠 Why This Is a Problem:

- In real-world use, we want to **predict `cnt` before the day ends** — before `casual` and `registered` are known.
- These two columns are **leakage variables** — they contain future or outcome-dependent information.
- Keeping them in the input means the model is cheating — it's learning from information it wouldn't have at prediction time.

### ✅ Best Practice:

- Drop `casual` and `registered` from the feature set (`X`) when training the model.
- Train using only those features that would be **known or measurable ahead of time** (e.g., weather, holiday, season, etc.)

```python
X = df_bikes.drop(columns=['cnt', 'casual', 'registered'])
y = df_bikes['cnt']
```

**Understanding regression**

### 📌 **Bike Rental Prediction is a Regression Problem**

Predicting the number of bike rentals involves estimating a **non-negative integer** value, which makes it a **regression task** in machine learning — where the target can take on a continuous range of values.

One of the most common algorithms for this is **linear regression**, which models the target as a weighted sum of input features (predictors). These weights (also called **coefficients**) are learned using an optimization technique like **gradient descent**, which minimizes prediction error. The output of linear regression can be **any real number (including negative values)**, even though the actual rental count is always a non-negative integer.


**How to Handle negative predictions?**

> 👉 **Linear regression predictions can be *any real number*, including negative values.**

---

### ❓ But Why Is That a Problem?

In many real-world problems — like predicting:

- **Bike rentals**
- **Housing prices**
- **Population counts**
- **Sales volume**

…the **target variable can’t be negative**. But since linear regression is a **mathematical model**, it doesn't "know" these domain-specific rules — it simply finds the **best-fit line**.

So if the inputs (features) push the prediction far enough, it **can return negative values**.

---

### 🛠 How to Handle This:

| Approach | Description |
|----------|-------------|
| **Post-processing** | Clamp predictions with `max(0, prediction)` to avoid negative outputs. |
| **Use models with non-negative constraints** | Some versions of regression (e.g., `sklearn.linear_model.Lasso` or `Ridge` with constraints) can limit weights. |
| **Log-transform the target** | Model `log(y)` instead of `y`, then apply `exp()` to get the final output. This ensures positive predictions. |
| **Use tree-based models** | Models like XGBoost or RandomForest naturally avoid unreasonable negatives in most regression tasks. |

**Predicting total number of bike rentals for each day**

````python
# --------------------------------------------------------
# Step 1: Import train_test_split function
# --------------------------------------------------------
# This utility is used to split the dataset into training and test subsets
# It helps evaluate how well the model generalizes to unseen data
from sklearn.model_selection import train_test_split

# --------------------------------------------------------
# Step 2: Import LinearRegression model class
# --------------------------------------------------------
# This is scikit-learn’s implementation of Ordinary Least Squares (OLS) linear regression
from sklearn.linear_model import LinearRegression

# --------------------------------------------------------
# Step 3: Split the dataset into training and test sets
# --------------------------------------------------------
# - X: feature matrix (input variables)
# - y: target vector (output variable)
# - test_size is not explicitly set, so default is 0.25 (25% test, 75% train)
# - random_state=2 ensures reproducibility (same split each time)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=2)
````

 Whenever you see random_state=2, this means that you are choosing the seed of a pseudo-random number generator to ensure reproducible results.

````python
# --------------------------------------------------------
# Step: Silence warnings in the notebook or script output
# --------------------------------------------------------

# Import the warnings module — part of Python's standard library
# It lets you control how warning messages are handled
import warnings

# Suppress all warning messages globally
# This is useful to hide non-critical warnings (e.g., deprecation or convergence warnings)
# Typically done for cleaner notebook outputs or demos
warnings.filterwarnings('ignore')
````

````python
# --------------------------------------------
# Step 1: Initialize the Linear Regression model
# --------------------------------------------
# Create an instance of scikit-learn's LinearRegression class
# This model will learn the relationship between features (X) and target (y)
lin_reg = LinearRegression()

# --------------------------------------------
# Step 2: Train the model on the training set
# --------------------------------------------
# Fit the linear regression model using the training data
# X_train: input features, y_train: corresponding labels
lin_reg.fit(X_train, y_train)

# --------------------------------------------
# Step 3: Predict on the test set
# --------------------------------------------
# Use the trained model to predict the target values for the test features (X_test)
y_pred = lin_reg.predict(X_test)

# --------------------------------------------
# Step 4: Evaluate the model using RMSE
# --------------------------------------------
# Import the mean_squared_error function from scikit-learn
from sklearn.metrics import mean_squared_error

# Import numpy for square root computation
import numpy as np

# Calculate Mean Squared Error (MSE) between actual and predicted test labels
mse = mean_squared_error(y_test, y_pred)

# Take square root of MSE to get Root Mean Squared Error (RMSE)
# RMSE gives a measure of how far predictions are from actual values on average
rmse = np.sqrt(mse)

# --------------------------------------------
# Step 5: Display the RMSE
# --------------------------------------------
# Print the RMSE rounded to 2 decimal places
print("RMSE: %0.2f" % (rmse))

# --------------------------------------------
# Step 6: Display R2 score
# --------------------------------------------
from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
print("R² Score: %0.4f" % r2)
````

## **Cross-validation**

**Cross-validation with linear regression**

````python
# -----------------------------------------------------------
# Step 1: Import the cross_val_score utility from sklearn
# -----------------------------------------------------------
# cross_val_score allows you to easily perform k-fold cross-validation
from sklearn.model_selection import cross_val_score

# -----------------------------------------------------------
# Step 2: Instantiate the regression model (Linear Regression)
# -----------------------------------------------------------
# This creates the model object we'll pass to cross_val_score
model = LinearRegression()

# -----------------------------------------------------------
# Step 3: Perform 10-fold cross-validation
# -----------------------------------------------------------
# - model: the regression algorithm to evaluate
# - X, y: input features and target variable
# - scoring='neg_mean_squared_error': sklearn returns negative MSE (so we need to invert it later)
# - cv=10: use 10-fold cross-validation
scores = cross_val_score(model, X, y, scoring='neg_mean_squared_error', cv=10)

# -----------------------------------------------------------
# Step 4: Convert negative MSE to positive RMSE
# -----------------------------------------------------------
# -scores converts to positive MSE
# np.sqrt() gives us RMSE (Root Mean Squared Error)
rmse = np.sqrt(-scores)

# -----------------------------------------------------------
# Step 5: Print RMSE for each of the 10 folds
# -----------------------------------------------------------
# Round each RMSE value to 2 decimal places for readability
print('Reg rmse:', np.round(rmse, 2))

# -----------------------------------------------------------
# Step 6: Print the average RMSE across all folds
# -----------------------------------------------------------
# This gives a more stable estimate of model performance
print('RMSE mean: %0.2f' % (rmse.mean()))
````

***Tip***

Why **scoring='neg_mean_squared_error'**?

Scikit-learn is designed to select the highest score when training models. This works well for accuracy, but not for errors when the lowest is best. By taking the negative of each mean squared error, the lowest ends up being the highest. This is compensated for later with rmse = np.sqrt(-scores), so the final results are positive.

**Cross-validation with XGBoost**

````python
# -----------------------------------------------------------
# Step 1: Instantiate the XGBoost Regressor
# -----------------------------------------------------------
# - We set the objective to "reg:squarederror" for regression tasks
# - This tells XGBoost to use squared error as the loss function
model = XGBRegressor(objective="reg:squarederror")

# -----------------------------------------------------------
# Step 2: Perform 10-fold cross-validation
# -----------------------------------------------------------
# - model: the regression algorithm to evaluate
# - X, y: input features and target variable
# - scoring='neg_mean_squared_error': returns negative MSE (standard in sklearn)
# - cv=10: perform 10 different train/test splits (folds)
scores = cross_val_score(model, X, y, scoring='neg_mean_squared_error', cv=10)

# -----------------------------------------------------------
# Step 3: Convert negative MSE to positive RMSE
# -----------------------------------------------------------
# -scores gives us positive MSE
# np.sqrt() converts it to RMSE (Root Mean Squared Error)
rmse = np.sqrt(-scores)

# -----------------------------------------------------------
# Step 4: Print RMSE for each fold
# -----------------------------------------------------------
# This shows the prediction error (in original units) for each test fold
print('Reg rmse:', np.round(rmse, 2))

# -----------------------------------------------------------
# Step 5: Print the average RMSE across all folds
# -----------------------------------------------------------
# This is a more stable estimate of model performance than any single test split
print('RMSE mean: %0.2f' % (rmse.mean()))
````

````python
# ------------------------------------------------------------
# Step: Split the dataset into features (X) and target (y)
# ------------------------------------------------------------

# X = all rows, and all columns except the last one (features)
# df.iloc[row_slice, column_slice]
# - iloc[:,:-1] means: for every row (:), select all columns except the last one (:-1)
# - This works assuming the last column is the target (e.g., 'income_>50K')
X = df_census.iloc[:, :-1]

# y = all rows, and only the last column (target variable)
# df.iloc[row_slice, column_slice]
# - iloc[:,-1] means: for every row (:), select only the last column (-1)
# - This gives us the binary target column we kept (e.g., 'income_>50K')
y = df_census.iloc[:, -1]
````

**Logistic regression**

````python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression
````

````python
# -------------------------------------------------------------
# Step 1: Import cross_val_score from scikit-learn
# -------------------------------------------------------------
# cross_val_score handles k-fold cross-validation behind the scenes
from sklearn.model_selection import cross_val_score

# -------------------------------------------------------------
# Step 2: Define a function to evaluate a classifier using CV
# -------------------------------------------------------------
# This function takes:
# - classifier: the model you want to evaluate (e.g., LogisticRegression)
# - num_splits: how many folds (default is 10 for 10-fold CV)
def cross_val(classifier, num_splits=10):

    # Step 2a: Initialize the classifier model
    model = classifier  # e.g., LogisticRegression()

    # Step 2b: Perform cross-validation and return accuracy scores
    # - cv=num_splits splits the data into k folds
    # - The model is trained and tested k times on different subsets
    scores = cross_val_score(model, X, y, cv=num_splits)

    # Step 2c: Print accuracy from each fold (rounded to 2 decimals)
    print('Accuracy:', np.round(scores, 2))

    # Step 2d: Print the average accuracy across all folds
    print('Accuracy mean: %0.2f' % (scores.mean()))

````

````python
# -------------------------------------------------------------
# Step 3: Call the function with LogisticRegression as classifier
# -------------------------------------------------------------
cross_val(LogisticRegression())
````

### ❓ Where Is the **Sigmoid Function** Used?

You don’t see it directly in the code, **but it’s built into the LogisticRegression model**:

```python
LogisticRegression()
```

Internally, during:
- **Training**: the model fits weights using **log loss (cross-entropy)**, which is based on the sigmoid function
- **Prediction**: it computes \( z = w^T x + b \), and applies:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

This sigmoid output gives the **probability of class 1**, which is then used to classify based on a threshold (usually 0.5).

````python
# -------------------------------------------------------------
# Use cross_val function to evaluate XGBoost Classifier
# -------------------------------------------------------------
# XGBClassifier is a powerful tree-based ensemble classifier from the XGBoost library
# - n_estimators=5 means the model will build 5 boosted decision trees
# - Each tree corrects the errors of the previous one (gradient boosting)

# Compared to LogisticRegression:
# - LogisticRegression is a **linear model** that uses the **sigmoid function**
#   to map the linear combination of features to a probability
# - XGBClassifier is a **nonlinear model** based on **decision trees**
#   and uses gradient boosting to improve performance iteratively
# - LogisticRegression has a simple decision boundary (linear)
# - XGBClassifier can learn complex, nonlinear boundaries and feature interactions

# This call performs 10-fold cross-validation by default
# and prints the accuracy per fold and the mean accuracy
cross_val(XGBClassifier(n_estimators=5))
````
