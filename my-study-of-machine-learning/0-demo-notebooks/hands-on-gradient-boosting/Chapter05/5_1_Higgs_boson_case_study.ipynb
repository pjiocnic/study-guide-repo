# Finding the Higgs boson – case study

The dataset **`atlas-higgs-challenge-2014-v2.csv.gz`**
is a very famous **benchmark dataset** from a public Kaggle competition:

👉 **Higgs Boson Machine Learning Challenge** (ATLAS experiment at CERN).

---

### 🚀 Background

* The **ATLAS experiment** is part of the Large Hadron Collider (LHC) at CERN.
* Physicists were trying to detect the **Higgs boson**, a fundamental particle predicted by the Standard Model of particle physics.
* The challenge: given **simulated collision event data**, build an ML model to distinguish:

| Class              | Meaning                                                              |
| ------------------ | -------------------------------------------------------------------- |
| `signal` → `s`     | A collision likely containing a Higgs boson signal                   |
| `background` → `b` | A collision containing only background noise (other known processes) |

---

### 📂 The dataset file

* CSV file, compressed with `.gz`
* Each row = 1 particle collision event (simulated data)
* Columns = physics features computed from raw collision data

---

To match the Kaggle training data, let's delete the Kaggleset and Weight columns, convert KaggleWeight into 'Weight', and move the 'Label' column to the last column

```python
# Delete the existing 'Weight' column from the DataFrame
# Why? → You are about to replace it with the correct 'KaggleWeight' column, renamed to 'Weight'
del df['Weight']

# Delete the 'KaggleSet' column from the DataFrame
# 'KaggleSet' indicates which rows belong to the training set or leaderboard sets (public/private).
# It is not useful for training the ML model itself, so we remove it.
del df['KaggleSet']

# Rename the 'KaggleWeight' column to 'Weight'
# 'KaggleWeight' is the correct event weight provided by Kaggle for model evaluation.
# We rename it to 'Weight' for consistency and easier access later.
df = df.rename(columns={"KaggleWeight": "Weight"})

# Temporarily store the 'Label' column (target values) into a separate variable.
# 'Label' contains the target: 's' (signal) or 'b' (background).
label_col = df['Label']

# Delete the original 'Label' column from the DataFrame.
# We'll add it back later as the last column, for convenience in viewing.
del df['Label']

# Add the 'Label' column back into the DataFrame as the last column.
# This is purely for readability — having the target column at the end is often helpful when exploring the data.
df['Label'] = label_col

# Display the first 5 rows of the cleaned DataFrame.
# You should now see:
# - Physics features columns
# - 'Weight' column (correct weights)
# - 'Label' column (target) as the last column
df.head()
```

## Higgs ML Dataset: Column Descriptions

---

### Column 0: `EventId`

- **Irrelevant for the machine learning model.**  
- Unique identifier for each event.  
- It is not a useful feature for training the model and can be dropped.

---

### Columns 1-30: Physics Columns

- **Derived from LHC collision data.**  
- These columns are computed physics quantities (such as invariant masses, transverse momenta, angular differences, etc.).  
- They serve as the **machine learning predictor columns (features)**.  
- For detailed descriptions of these columns, refer to the [technical documentation](http://higgsml.lal.in2p3.fr/documentation).

---

### Column 31: `Weight`

- **Event weight used to scale the data during evaluation.**
- Why is this important?
  - Higgs boson events are **extremely rare** compared to background events.
  - Without weighting, a model could achieve **high accuracy** just by predicting "background" all the time (because background dominates the data).
  - Weights help correct this **class imbalance** by giving rare signal events more "importance" during training and evaluation.
  - ⚠️ **Note:** Weights are not provided in the test set used for submission — special care is required to handle this correctly.
  - Strategies for dealing with weights are covered later (and further in *Chapter 7: Discovering Exoplanets with XGBoost*).

---

### Column 32: `Label`

- **Target column for supervised learning.**
- Values:
  - `'s'` → **signal** (event likely containing Higgs boson decay)
  - `'b'` → **background** (other physics processes — not Higgs decay)
- The training data is **simulated** based on real LHC data:
  - There are **more "signal" events in training than would be found in r**


### Preprocessing

The only issue with the data is that the target column, Label, is not numerical.

Convert the Label column into a numerical column by replacing the s values with 1 and the b values with 0

```python
pd.set_option('future.no_silent_downcasting', True)
df['Label'].replace(('s', 'b'), (1, 0), inplace=True)
```
