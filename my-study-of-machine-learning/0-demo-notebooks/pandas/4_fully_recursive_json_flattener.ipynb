This notebook recursively flattens all nested dicts and list-of-dicts into a single row per JSON.

```python
import os
import json

# Set your directory path
json_dir = "data/"

# List all JSON files in the directory
json_files = [f for f in os.listdir(json_dir) if f.endswith(".json")]

# Load each JSON file
json_data = []
for fname in json_files:
    full_path = os.path.join(json_dir, fname)
    with open(full_path, 'r') as f:
        json_data.append(json.load(f))

print(f"Loaded {len(json_data)} JSON records.")
```

## 🔄 Recursive Flattener Function

```python
import pandas as pd

def flatten_item(prefix, item):
    """
    Recursively flattens a JSON item (dict or list) with a given prefix.
    Returns a flat dict.
    """
    flattened = {}
    if isinstance(item, dict):
        for k, v in item.items():
            new_prefix = f"{prefix}.{k}" if prefix else k
            flattened.update(flatten_item(new_prefix, v))
    elif isinstance(item, list):
        for i, elem in enumerate(item):
            new_prefix = f"{prefix}[{i}]"
            flattened.update(flatten_item(new_prefix, elem))
    else:
        flattened[prefix] = item
    return flattened

def fully_flatten_json_records(json_list):
    """
    Flattens a list of deeply nested JSON records.
    Returns a DataFrame with one row per record and fully expanded keys.
    """
    flat_records = []
    for record in json_list:
        flat_dict = flatten_item("", record)
        flat_records.append(flat_dict)
    return pd.DataFrame(flat_records)
```

## 📊 Flatten and Inspect

```python
df_flat = fully_flatten_json_records(json_data)
print("✅ Flattened shape:", df_flat.shape)
df_flat.head()
```
