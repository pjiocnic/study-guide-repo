## 📚 What Are Checkpoint Folders?


When you train using Hugging Face `Trainer`, it saves model checkpoints automatically after certain steps (like every 500 steps).

- **Folder names**: `checkpoint-500/`, `checkpoint-1000/`, etc.
- **Purpose**: Allows you to resume training if interrupted.
- **Contents**: Each checkpoint has model weights and configs.

✅ **For final inference**, you **only need the final model** saved by `trainer.save_model()`.  
✅ **You can safely exclude** checkpoint folders from your ZIP to save space!


## 📂 Files Inside Model Folder and Their Purpose


| File Name | Purpose | Needed for Inference? |
|:--|:--|:--|
| **pytorch_model.bin** | 🧠 Model weights (learned parameters) | ✅ **Required** |
| **config.json** | 🛠️ Model architecture and task configuration | ✅ **Required** |
| **tokenizer_config.json** | ⚙️ Tokenizer settings (e.g., lower casing) | ✅ **Required** |
| **vocab.txt** | 📖 Vocabulary used by tokenizer | ✅ **Required** |
| **tokenizer.json** | 🗂️ Serialized tokenizer structure | ✅ **Required** |
| **special_tokens_map.json** | 🎯 Mapping of special tokens ([CLS], [SEP], etc.) | ✅ **Required** |
| **added_tokens.json** (optional) | ➕ Custom tokens added (if any) | ❓ Optional |
| **trainer_state.json** | 📈 Training statistics and history | ❌ **Not needed** |
| **checkpoint-*** folders | 🏗️ Intermediate saved models | ❌ **Not needed** |

✅ In production or for inference, **only keep core model and tokenizer files**.


## 🛠 Create a Clean ZIP Excluding Checkpoints

```

import os
import zipfile

def zip_model_without_checkpoints(model_folder, zip_filename):
    with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(model_folder):
            # Skip any checkpoint folders
            if any("checkpoint" in d for d in root.split(os.sep)):
                continue
            for file in files:
                file_path = os.path.join(root, file)
                arcname = os.path.relpath(file_path, model_folder)
                zipf.write(file_path, arcname)

# Example usage
model_folder = "./ner_model"  # Your saved model folder
zip_filename = "ner_model_clean.zip"

zip_model_without_checkpoints(model_folder, zip_filename)

print(f"✅ Clean ZIP created: {zip_filename}")
```

## 📂 Unzip the Uploaded Model

```

import zipfile

# Unzip
with zipfile.ZipFile("ner_model_clean.zip", 'r') as zip_ref:
    zip_ref.extractall("./ner_model")

print("✅ Model unzipped successfully!")
```
