# 📘 Regressors vs. Classifiers — Study Guide with Bike Rentals Example

## ✅ What Are They?
| Term | Definition |
|------|------------|
| **Regressor** | A model that predicts **continuous numerical values** (e.g., price, temperature, rentals). |
| **Classifier** | A model that predicts **categories or class labels** (e.g., spam/ham, yes/no). |

## 📊 Real Example: Bike Rentals (Regression)
We'll predict the **number of bike rentals** (`cnt`) using features like weather, season, and weekday.

### 🔽 Dataset: `bike_rentals.csv`
Ensure your dataset includes:
- `cnt` (target)
- `temp`, `atemp`, `hum`, `windspeed`, `season`, `weekday`, `holiday`, etc.

```python
# 📦 Load data and prepare environment
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Load dataset
df = pd.read_csv("bike_rentals.csv")

# Drop known leakage columns
X = df.drop(columns=['cnt', 'casual', 'registered'])
y = df['cnt']

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=2)
```

```python
# 🤖 Train and evaluate Linear Regression
model = LinearRegression()
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
rmse = mean_squared_error(y_test, y_pred, squared=False)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

## 🔍 Output Meaning
- **RMSE** tells how far off your predictions are (in rental counts).
- **R² Score** tells how much of the variation in rentals your model explains.

Try swapping in other regressors (e.g., `XGBRegressor`, `RandomForestRegressor`) and compare results!

# 🛳️ Classifier Example: Titanic Survival Prediction
We'll use the Titanic dataset to classify whether a passenger **survived** or not (1 or 0).

```python
# 📦 Load Titanic dataset and preprocess
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load data
titanic = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

# Select features and drop missing
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
titanic = titanic[features + ['Survived']].dropna()
titanic['Sex'] = titanic['Sex'].map({'male': 0, 'female': 1})

X = titanic[features]
y = titanic['Survived']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=2)
```

```python
# 🤖 Train a classifier
clf = RandomForestClassifier()
clf.fit(X_train, y_train)

# Predict and evaluate
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
```

## 🔍 Output Meaning
- **Accuracy** tells what % of passengers were correctly classified as survived or not.
- This is a classification task because the target (`Survived`) is categorical (0 or 1).

# 📊 Visualizing Model Performance
We'll compare predicted vs. actual values for regression and plot the confusion matrix for classification.

```python
# 📈 Regression: Actual vs Predicted
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel('Actual Bike Rentals')
plt.ylabel('Predicted Bike Rentals')
plt.title('Linear Regression: Actual vs Predicted')
plt.grid(True)
plt.show()
```

```python
# 📉 Classification: Confusion Matrix
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay.from_estimator(clf, X_test, y_test)
plt.title('Random Forest Classifier: Confusion Matrix')
plt.grid(False)
plt.show()
```

## 🔍 Classification: ROC Curve & Feature Importance
We'll visualize the ROC curve to evaluate classifier performance and inspect which features mattered most.

```python
# 📊 ROC Curve for Titanic Classifier
from sklearn.metrics import RocCurveDisplay
RocCurveDisplay.from_estimator(clf, X_test, y_test)
plt.title('Random Forest Classifier: ROC Curve')
plt.grid(True)
plt.show()
```

```python
# 📊 Feature Importance Plot
import numpy as np

importances = clf.feature_importances_
indices = np.argsort(importances)[::-1]
feature_names = X.columns

plt.figure(figsize=(8, 6))
plt.bar(range(len(importances)), importances[indices], align='center')
plt.xticks(range(len(importances)), feature_names[indices], rotation=45)
plt.title('Feature Importances in Random Forest')
plt.tight_layout()
plt.show()
```
