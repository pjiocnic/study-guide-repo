# 📚 Import Libraries

```

from transformers import pipeline
from datasets import load_dataset
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
```

# 📥 Load TweetEval Sentiment Dataset

```

# Load only the 'sentiment' subset
dataset = load_dataset("tweet_eval", "sentiment")
dataset
```

# 👀 Explore Dataset Examples

```

print("Example tweet:", dataset['test'][0]['text'])
print("Label (0 = negative, 1 = neutral, 2 = positive):", dataset['test'][0]['label'])
```

# 🤖 Load DistilBERT Sentiment Model

```

sentiment_pipeline = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
```

# 🔮 Predict Sentiments for Sample Tweets

```

# Predict for a small batch
texts = [dataset['test'][i]['text'] for i in range(50)]  # take 50 tweets
true_labels = [dataset['test'][i]['label'] for i in range(50)]

# Model predicts only POSITIVE or NEGATIVE (binary), so we'll map neutral to closest
predictions = sentiment_pipeline(texts)

# Map model outputs to numerical labels
label_map = {'NEGATIVE': 0, 'POSITIVE': 2}
predicted_labels = [label_map[p['label']] for p in predictions]
```

# 📊 Evaluate Accuracy

```

# Ignore NEUTRAL labels in ground truth for fair comparison
filtered_true = []
filtered_pred = []

for t, p in zip(true_labels, predicted_labels):
    if t != 1:  # Ignore neutral
        filtered_true.append(t)
        filtered_pred.append(p)

acc = accuracy_score(filtered_true, filtered_pred)
print(f"Accuracy (ignoring neutral tweets): {acc:.2f}")
```

# 📈 Visualize Sentiment Distribution

```

# Count positive and negative predictions
from collections import Counter

counter = Counter(predicted_labels)
labels = ['Negative (0)', 'Positive (2)']
counts = [counter.get(0, 0), counter.get(2, 0)]

plt.bar(labels, counts, color=['red', 'green'])
plt.title("Predicted Sentiment Distribution")
plt.ylabel("Count")
plt.show()
```
