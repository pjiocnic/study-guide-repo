# **Exploring decision trees**

**First decision tree model**

We start by building a decision tree to predict **whether someone makes over 50K** US dollars using the Census dataset

**<u>The process</u>**

As with other machine learning classifiers, when using decision trees, we initialize the model, fit it on the training set, and **test** it using **accuracy_score**.

**<u>How is accuracy determined?</u>**

The accuracy_score determines the number of correct predictions divided by the total number of predictions.
If 19 of 20 predictions are correct, the accuracy_score is 95%.

**Note:** you are comparing **y_pred** to **y_test** - `accuracy_score(y_pred, y_test)`


````python
# Import pandas and numpy
import pandas as pd
import numpy as np

# Import warnings
import warnings
warnings.filterwarnings('ignore')
````

````python
# 📥 Load the cleaned census dataset from CSV file
df_census = pd.read_csv('census_cleaned.csv')  # Assumes the last column is the target variable (label)
df_census.head()
````

````python
# 🧪 Split the dataset into features (X) and target (y)

# X: all rows, all columns except the last one (features)
# iloc[:, :-1] means “select all rows (:) and all columns up to but not including the last one (:-1)”
X = df_census.iloc[:, :-1]

# y: all rows, only the last column (target label)
# iloc[:, -1] means “select all rows (:) and only the last column (-1)”
y = df_census.iloc[:, -1]

# 🧪 Import train_test_split to partition data into training and testing sets
from sklearn.model_selection import train_test_split

# 📊 Split the dataset into training and test sets
# By default, 75% of the data goes to training and 25% to testing
# random_state=2 ensures reproducibility: same split every time this code runs
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=2)
````

````python
# 🌳 Import the Decision Tree Classifier from scikit-learn
from sklearn.tree import DecisionTreeClassifier

# 🎯 Import accuracy_score to evaluate prediction performance
from sklearn.metrics import accuracy_score

# 🧠 Initialize the Decision Tree classifier
# Setting random_state=2 ensures consistent and repeatable behavior across runs,
# especially when the tree has choices (e.g., which feature to split on when multiple have equal scores)
clf = DecisionTreeClassifier(random_state=2)

# 🏋️ Fit (train) the classifier on the training data
# The model learns patterns from X_train and their corresponding labels y_train
clf.fit(X_train, y_train)

# 🔮 Make predictions using the trained model on unseen test data
# This returns an array of predicted labels for each row in X_test
y_pred = clf.predict(X_test)

# 📏 Calculate and return the accuracy of the predictions
# Compares predicted labels (y_pred) to true labels (y_test)
# Accuracy = (Number of correct predictions) / (Total predictions)
accuracy_score(y_pred, y_test)
````

An accuracy of 81% is **comparable** to the accuracy of **Logistic Regression** from the same dataset

````python
# 🚲 Load the cleaned bike rentals dataset
df_bikes = pd.read_csv('bike_rentals_cleaned.csv')  # Assumes the last column is the target (e.g., number of rentals)

# 🧪 Split the dataset into features (X_bikes) and target (y_bikes)

# X_bikes: all rows, all columns except the last one
# iloc[:, :-1] selects all rows (:) and all columns up to but not including the last one
X_bikes = df_bikes.iloc[:, :-1]

# y_bikes: all rows, only the last column (target variable)
# iloc[:, -1] selects the last column of every row
y_bikes = df_bikes.iloc[:, -1]

# 📈 Import the Linear Regression model from scikit-learn
from sklearn.linear_model import LinearRegression

# 📊 Split the dataset into training and testing sets
# 75% of the data goes to training and 25% to testing by default
# random_state=2 ensures reproducibility—the same rows will be in train/test each time you run it
X_train, X_test, y_train, y_test = train_test_split(X_bikes, y_bikes, random_state=2)
````

````python
# Import Decision Tree Regressor
from sklearn.tree import DecisionTreeRegressor

# Import cross_val_score
from sklearn.model_selection import cross_val_score
````

````python
# 🌳 Import the Decision Tree Regressor
from sklearn.tree import DecisionTreeRegressor

# 📏 Import cross_val_score for cross-validation
from sklearn.model_selection import cross_val_score

# 🧮 Import numpy for mathematical operations
import numpy as np

# 🧠 Initialize the Decision Tree Regressor
# random_state=2 ensures the splits in the tree are consistent across runs
reg = DecisionTreeRegressor(random_state=2)

# 🔁 Perform 5-fold cross-validation on the dataset
# - scoring='neg_mean_squared_error' because cross_val_score expects higher scores to be better
# - So MSE (which is a loss function) is negated before scoring
# - cv=5 means the data is split into 5 folds (each fold is used once as a test set)
scores = cross_val_score(reg, X_bikes, y_bikes, scoring='neg_mean_squared_error', cv=5)

# 📉 Convert negative MSE to positive, then take the square root to get RMSE (Root Mean Squared Error)
rmse = np.sqrt(-scores)

# 📊 Display the average RMSE across all 5 folds
print('RMSE mean: %0.2f' % (rmse.mean()))
````

````python
# 🌳 Initialize the Decision Tree Regressor (no random_state set = non-deterministic behavior)
reg = DecisionTreeRegressor()

# 🏋️ Train (fit) the model on the training data
# The model learns patterns from X_train and y_train
reg.fit(X_train, y_train)

# 🔮 Predict the target values using the trained model on the same training set
# Note: Using training data here instead of test data → will likely result in low error
y_pred = reg.predict(X_train)

# 📏 Import mean_squared_error to compute regression loss
from sklearn.metrics import mean_squared_error

# 🧮 Calculate Mean Squared Error (MSE) between actual and predicted values on training set
reg_mse = mean_squared_error(y_train, y_pred)

# ✅ Calculate Root Mean Squared Error (RMSE)
# RMSE is the square root of MSE and is in the same unit as the target variable (e.g., bike rentals)
reg_rmse = np.sqrt(reg_mse)

# 🔢 Output RMSE value for interpretation
reg_rmse
````

````python
# 🔍 Import GridSearchCV for hyperparameter tuning
from sklearn.model_selection import GridSearchCV

# 🛠 Define hyperparameters to search over
# max_depth controls how deep the tree can grow
# Deeper trees can overfit, shallower trees can underfit
params = {'max_depth': [None, 2, 3, 4, 6, 8, 10, 20]}

# 🌳 Initialize a base Decision Tree Regressor
# random_state=2 ensures consistent results during tree construction
reg = DecisionTreeRegressor(random_state=2)

# 🔄 Initialize GridSearchCV for systematic hyperparameter tuning
# - reg: the model to tune
# - params: dictionary of hyperparameters to try
# - scoring: uses negative mean squared error (MSE) as metric (must be negated for maximization logic)
# - cv=5: uses 5-fold cross-validation to evaluate each setting
# - n_jobs=-1: utilizes all available CPU cores for faster search
grid_reg = GridSearchCV(reg, params, scoring='neg_mean_squared_error', cv=5, n_jobs=-1)

# 🏋️ Fit GridSearchCV on the training data
# This will train and evaluate multiple models with different max_depths
grid_reg.fit(X_train, y_train)

# 🏆 Extract the best-performing hyperparameters
best_params = grid_reg.best_params_

# 📢 Display the best hyperparameters found by grid search
print("Best params:", best_params)
````

````python
# ✅ Compute the best (lowest) RMSE from GridSearchCV

# grid_reg.best_score_ gives the **best cross-validation score** achieved
# Since we used scoring='neg_mean_squared_error', the value is negative
# We negate it back and take the square root to get RMSE (in original units of target)
best_score = np.sqrt(-grid_reg.best_score_)

# 🖨️ Print the best cross-validated training RMSE score
print("Training score: {:.3f}".format(best_score))
````

````python
# 🏆 Extract the best model from GridSearchCV
# This is the DecisionTreeRegressor with the best max_depth found via cross-validation
best_model = grid_reg.best_estimator_

# 🔮 Use the best model to make predictions on the test data
# These are the model's estimates for y_test
y_pred = best_model.predict(X_test)

# 📏 Import the mean_squared_error function
# This will be used to compute the RMSE (Root Mean Squared Error)
from sklearn.metrics import mean_squared_error

# 🧮 Compute RMSE on the test set
# First compute MSE, then take square root to convert to RMSE
rmse_test = mean_squared_error(y_test, y_pred) ** 0.5

# 📊 Print the test RMSE score
# This tells us how well the model performs on unseen data
print('Test score: {:.3f}'.format(rmse_test))
````

````python
# 🧪 Define a reusable function to perform grid search, training, and evaluation
def grid_search(params, reg=DecisionTreeRegressor(random_state=2)):

    # 🔍 Step 1: Set up GridSearchCV with cross-validation
    # - reg: the model to tune (default = DecisionTreeRegressor with fixed random_state)
    # - params: dictionary of hyperparameters to try (e.g., max_depth)
    # - scoring: uses negative MSE, which will be later converted to RMSE
    # - cv=5: 5-fold cross-validation
    # - n_jobs=-1: parallel processing on all CPU cores
    grid_reg = GridSearchCV(reg, params, scoring='neg_mean_squared_error', cv=5, n_jobs=-1)

    # 🏋️ Step 2: Fit the grid search object to training data
    grid_reg.fit(X_train, y_train)

    # 🏆 Step 3: Extract and display best hyperparameters
    best_params = grid_reg.best_params_
    print("Best params:", best_params)

    # 📊 Step 4: Compute and display training RMSE (cross-validated)
    # Negate best_score_ to get MSE, then take sqrt for RMSE
    best_score = np.sqrt(-grid_reg.best_score_)
    print("Training score: {:.3f}".format(best_score))

    # 🔮 Step 5: Make predictions on the test set using the best model found
    y_pred = grid_reg.predict(X_test)

    # 📏 Step 6: Compute RMSE on test data
    rmse_test = mean_squared_error(y_test, y_pred) ** 0.5
    print('Test score: {:.3f}'.format(rmse_test))
````

````python
# 🧪 Run grid_search on DecisionTreeRegressor with varying `min_samples_leaf` values

# min_samples_leaf: Minimum number of samples required to form a leaf node
# - Smaller values (e.g., 1) allow the tree to grow deep and capture noise → risk of overfitting
# - Larger values (e.g., 20, 30) make the tree simpler → better generalization, may increase bias

# This grid search will:
# 1. Train a model for each value in the list using 5-fold cross-validation
# 2. Score each model using negative mean squared error (MSE)
# 3. Select the value of `min_samples_leaf` that yields the lowest RMSE
# 4. Report the training RMSE (cross-validated) and test RMSE

grid_search(params={'min_samples_leaf': [1, 2, 4, 6, 8, 10, 20, 30]})
````

# **🧭 *SIDEBAR***

When you run:

```python
grid_search(params={'min_samples_leaf':[1,2,4,6,8,10,20,30]})
```

### 🔍 What This Does:

- You're tuning the **`min_samples_leaf`** hyperparameter of `DecisionTreeRegressor`.

---

### 🧠 What is `min_samples_leaf`?

| Parameter         | Meaning |
|------------------|---------|
| `min_samples_leaf` | The **minimum number of samples required to be in a leaf node** (i.e., a terminal node of the tree) |

---

### 🔧 Why Tune This?

- **Lower values (e.g., 1)** allow very **fine-grained splits**, increasing model complexity and risk of **overfitting**.
- **Higher values (e.g., 20, 30)** smooth the tree, leading to **simpler models**, potentially reducing variance and improving generalization.

---

### ✅ Expected Behavior of Grid Search:

- It will train and evaluate a different tree for each `min_samples_leaf` value.
- For each value, it performs **5-fold cross-validation** and selects the one with the lowest average **RMSE**.
- It will print:
  - Best `min_samples_leaf`
  - Cross-validated **training RMSE**
  - **Test RMSE** on unseen data

````python
# 🧪 Run grid_search to tune TWO hyperparameters of DecisionTreeRegressor:
#    - max_depth: maximum depth of the decision tree
#    - min_samples_leaf: minimum number of samples required to be in a leaf node

# 📌 Tuning both parameters simultaneously allows us to control:
#    - Model complexity (via `max_depth`)
#    - Granularity of splits (via `min_samples_leaf`)

# 💡 Why tune these?
# - `max_depth=None`: no limit on depth → risk of overfitting
# - `max_depth=2 to 20`: progressively allow deeper trees → more expressive models
# - `min_samples_leaf=1`: very fine splits → may capture noise
# - `min_samples_leaf=30`: smoother trees → potentially better generalization

# 🔄 GridSearchCV will:
# 1. Train one model for each combination of `max_depth` × `min_samples_leaf`
#    → total models = 8 depths × 8 leaf sizes = 64 models
# 2. Use 5-fold cross-validation to evaluate each model
# 3. Score them using negative mean squared error (converted to RMSE)
# 4. Return:
#    - Best hyperparameter combination
#    - Best training score (mean CV RMSE)
#    - Test RMSE for model selected via cross-validation

grid_search(params={
    'max_depth': [None, 2, 3, 4, 6, 8, 10, 20],
    'min_samples_leaf': [1, 2, 4, 6, 8, 10, 20, 30]
})
````

# **🧭 *SIDEBAR***

### 🧠 Summary of What Happens Internally:
| Step | Description |
|------|-------------|
| 1️⃣   | 64 combinations (8 × 8) are tested |
| 2️⃣   | 5-fold CV used for each → 320 total fits |
| 3️⃣   | Best combination found using lowest average RMSE |
| 4️⃣   | Test performance is reported on `X_test`, `y_test` |

````python
# 🧪 Run grid_search to fine-tune the DecisionTreeRegressor with a focused hyperparameter grid:
#    - max_depth: maximum allowed depth of the decision tree
#    - min_samples_leaf: minimum number of samples required in a leaf node

# 🎯 This refined search space is used for fine-tuning:
# - `max_depth = [5, 6, 7, 8, 9]`:
#     Controls how deep the tree can go (complexity).
#     - Lower values reduce overfitting, might underfit
#     - Higher values increase flexibility, risk overfitting
#
# - `min_samples_leaf = [3, 5, 7, 9]`:
#     Controls the minimum number of samples in terminal nodes (leaf nodes).
#     - Higher values create simpler, more general trees
#     - Lower values allow more detailed splits

# 🔄 GridSearchCV will:
# 1. Train 5 × 4 = 20 different tree configurations (each combination of parameters)
# 2. Use 5-fold cross-validation (total 100 training fits)
# 3. Evaluate using negative mean squared error (MSE), then report best RMSE
# 4. Predict and evaluate performance on the held-out test set

grid_search(params={
    'max_depth': [5, 6, 7, 8, 9],
    'min_samples_leaf': [3, 5, 7, 9]
})
````
