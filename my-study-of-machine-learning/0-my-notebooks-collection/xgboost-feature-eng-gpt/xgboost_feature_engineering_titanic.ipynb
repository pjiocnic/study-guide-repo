# 🚀 XGBoost Feature Engineering on Titanic-like Dataset
This notebook demonstrates:
- One-hot encoding
- Missing value imputation
- Date/time feature extraction
- XGBoost training
- Feature importance visualization

```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("titanic_raw.csv", parse_dates=['oracle_timestamp'])

# Fill missing age with median
df['age'] = df['age'].fillna(df['age'].median())

# Feature extraction from oracle_timestamp
df['hour'] = df['oracle_timestamp'].dt.hour
df['weekday'] = df['oracle_timestamp'].dt.weekday

# Drop unneeded columns (e.g., names)
df = df.drop(columns=['first_name', 'last_name', 'oracle_timestamp', 'oracle_date'])

# One-hot encode categorical features
df = pd.get_dummies(df, columns=['sex', 'embarked', 'country'], drop_first=True)

# Define features and target
X = df.drop(columns=['survived'])
y = df['survived']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42)

# Train XGBoost
model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Plot feature importance
importances = model.feature_importances_
features = X.columns
plt.figure(figsize=(10,6))
plt.barh(features, importances)
plt.xlabel("Importance")
plt.title("Feature Importance from XGBoost")
plt.tight_layout()
plt.show()
```
