# 🧹 Data Cleaning in Python – Colab Demo

This notebook demonstrates how to clean a messy dataset step-by-step using `pandas`.

```
import pandas as pd
import numpy as np

# Load the dataset
url = '/content/dirty_dataset.csv'
df = pd.read_csv(url)
df
```

## Step 1: Handle Missing Values

```
# Check missing values
df.isnull().sum()
```

```
# Drop rows with too many missing values
df = df.dropna(thresh=3)
df
```

## Step 2: Standardize Categorical Columns

```
# Fix casing and duplicates in Name
df['Name'] = df['Name'].str.strip().str.title()
df
```

## Step 3: Convert Age to Numeric

```
# Coerce non-numeric to NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df
```

## Step 4: Validate and Clean Email

```
df['Email'] = df['Email'].str.lower()
df['Email_Valid'] = df['Email'].str.contains(r'^[\w\.-]+@[\w\.-]+\.\w+$', na=False)
df
```

## Step 5: Parse Dates

```
df['Signup Date'] = pd.to_datetime(df['Signup Date'], errors='coerce')
df
```

## Step 6: Clean Numerical Scores

```
df['Score'] = pd.to_numeric(df['Score'], errors='coerce')
df['Score'] = df['Score'].fillna(df['Score'].median())
df
```
